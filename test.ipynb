### git과 github란?
- git은 내 컴퓨터(로컬)에서 파일의 변경 이력을 관리하는 버전 관리 시스템
- github는 git으로 관리하는 로컬 저장소를 원격 저장소로 보관하고 공유할 수 있는 서비스


### 1. Git으로 프로젝트 시작하기

`git init`

이미 내 컴퓨터(로컬)에 존재하는 프로젝트 폴더를 새로운 git 로컬 저장소로 지정하겠다는 의미이며, 지정을 하면 해당 폴더에 .git 디렉터리가 생성된다.
하지만 init으로만은 Github와 연결되는 것이 아니다. Github Repository와 연결하기 위해서 `remote` 설정이 필요하다.


`git clone`

Github에 이미 존재하는 Repository를 내 컴퓨터(로컬)에 복제하여 로컬 저장소를 만드는 명령어이다.
`clone`을 실행하면 파일만 복제하는 것이 아니라 Git정보와 원격 저장소 연결도 함께 설정된다.

따라서, 

이미 프로젝트 폴더가 로컬에 존재한다면 -> init

원격저장소에 프로젝트 파일이 존재하면 -> clone 


### 2. 기본적인 작업 흐름

파일 수정 -> git status -> git add -> git commit -> git push

`git status`
현재 Repository를 확인한다. 어떤 파일이 수정되었는지, `add`된 파일이 무엇인지, 현재 브랜치는 무엇인지 등을 확인할 수 있다.
파일을 변경하는 명령어가 아니기 때문에 작업 중 수시로 사용해도 괜찮다.

`git add`
작업된 파일을 staging Area로 올리는 과정이다.

`git commit`
staging Area에 올라온 변경사항을 하나의 버전으로 로컬 Repository에 기록한다.

`git push`
로컬 Repository에 저장된 Commit을 Github의 원격 Repository로 전송한다.
